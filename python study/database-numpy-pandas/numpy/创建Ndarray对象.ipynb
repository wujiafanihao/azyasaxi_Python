```python
import numpy as np
```

```python
arr = np.array([1,2,3,4,5,6],ndmin=2)
arr
```

```python
arr = np.array([3,2],dtype=int)

print(arr)
```

```python
arr = np.zeros(5,dtype=int)
arr
```

```python
arr = np.zeros([3,2],dtype=int)
arr
```

```python
arr = np.zeros([2,2],dtype=[('x','i4'),('y','f')])
arr
```

```python
arr = np.ones([3,2],dtype='i4')
arr
```

```python
arr = np.full([3,2],fill_value=1024,dtype='i4')
arr
```

```python
arr = np.arange(1, 11, 2)
print(arr)
```

```python
s = b"wujiafa zuishuai"
arr = np.frombuffer(s,dtype='S1',count=5,offset=2)
arr
```

```python
#从迭代对象创建ndarray
x = [1,2,3,4,5,6]
z = iter(x)
print(z)
print(type(z))
arr = np.fromiter(z,dtype='f')
arr
```

```python
#创建等差数列
arr = np.linspace(1,10,10,dtype='i4')
arr
```

```python
#创建等比数列
arr = np.logspace(1.0,2.0,10,dtype='f4')
arr
```
