```python
import numpy as np

a = np.arange(12)
a.shape = (3,4)
print(a)
print('-----------------------')
for x in np.nditer(a):
    print(x,end=' ')
print(" ")
```

```python
b = a.T.copy(order = 'F') #C为行优先 F为列优先
print(b)
print('----------------------')
for x in np.nditer(b):
    print(x,end=",")
print("")
```

```python
b = a.T.copy(order = 'F') #C为行优先 F为列优先
print(b)
print('----------------------')
for x in np.nditer(b,order = 'C', op_flags = ["readwrite"]):
    x[...] = x * 2
print(b)
```
