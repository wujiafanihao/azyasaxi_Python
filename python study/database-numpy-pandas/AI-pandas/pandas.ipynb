## 1.3数据的输入和输出

#### 1.3.1csv

```python
import numpy as np
import pandas as pd

# 薪资情况
df = pd.DataFrame(data = np.random.randint(5000,30000,(50,5)),columns=['IT','化工','生物','教师','士兵',]) 
df
```

```python
# 保存csv
df.to_csv(
    './salary.csv', # 起名 
    # sep=';', # 文本分隔符,默认逗号
    header = True, # 是否保存列索引
    index = True # 是否保存行索引
)
df
```

```python
# 加载
df = pd.read_csv(
    './salary.csv',
    sep=';',
    header=[0],
    index_col=0
)
df
```

```python
import numpy as np
import pandas as pd

# 薪资情况
df1 = pd.DataFrame(data = np.random.randint(5000,30000,(50,5)),columns=['IT','化工','生物','教师','士兵',]) 
df2 = pd.DataFrame(data = np.random.randint(0,50,size=(150,3)),columns=['Python','Math','English'])

display(df1,df2)
```

```python
df1.to_excel(
    './salary.xlsx',
    header = True,
    index = False,
)
df1
```

```python
df = pd.read_excel(
    './salary.xlsx',
    header=0,
    names=list('ABCDE'), # 替换列索引
    index_col=0 # 指定汉元素，A作为行索引
)
df
```

```python
# 一个excel文件中保存多个工作表
with pd.ExcelWriter('./data.xlsx') as writer:
    df1.to_excel(writer,sheet_name='salary',index=False)
    df2.to_excel(writer,sheet_name='exam',index=False)
pd.read_excel(
    './data.xlsx',
    sheet_name='salary'
)
```
