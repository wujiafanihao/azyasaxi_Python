```python
import numpy as np
```

```python
arr = [1,2,3,4,5]
print('print:',arr)

print('print 1 database:',arr[0])

print('print last database:',arr[4])
print('print last database:',arr[-1])
```

```python
arr_np = np.array(arr)
arr_np
```

```python
print('print 1 and 3 database : ',arr_np[[0,2]])

print('print 1 and 3 database : ',arr_np[[0,2]])
arr_np[[0,2]]
```

```python
#打印大于3的
cond = arr_np >= 3
cond
arr_np[cond]
```

# **Numpy切片和索引**

## **一位数组的索引的切片**

```python
arr = np.random.randint(0,100,size=10)
arr
```

```python
print(arr[0]) # 索引
print(arr[-1]) # 索引
print(arr[0:2]) # 切片
print(arr[[0,3,5]]) # 索引
```

```python
print(arr[3:6])
print(arr[3:6:2]) #从3-6，2表示两个中取一个
```

```python
arr[::2] # 每隔2个取1个
```

```python
arr[::-1] # 逆序
```

```python
arr[::-3] # 逆序 每隔3个取1个
```

## **二维数组的索引和切片**

```python
arr2 = np.random.randint(0,100,size = (3,5))
arr2
```

```python
# 逗号隔开 取97 第一行第三列
arr2[0,2]
```

```python
arr2[1,[1,2]] #第二行的第二第三个数据
```

```python
# 第二行和第三行以及第四列和第五列的数据
# 索引和切片配合使用
arr2[[1,2],3:]
```

```python
arr2[1::-1,[0,1]] # 先降序，然后取二三行的 一二列 
```

```python
arr2[[1,2]][:,[3,4]] # 取二三行的四五列
```

```python
arr2[:][:,[0,2,4]] # 先默认,然后再取一三五列
```

```python
arr2[2,2] = 1024
arr2 #找到索引改值
```

```python
arr2[1,2:4] = 2048
arr2
```

```python
arr2[0,[1,4]] = 8096
arr2
```

## **花式索引**

```python
arr3 = np.random.randint(0,100,size = 20)
arr3
```

```python
arr3[[3,5,7]] # 花式索引
```

```python
cond = arr3 > 60
arr3[cond] # 取出大于60的数据
```

```python
# 取出小于30,大于80的数据
cond1 = arr3 > 80
cond2 = arr3 < 30

cond = cond1 | cond2 # 或运算 

arr3[cond]
```
