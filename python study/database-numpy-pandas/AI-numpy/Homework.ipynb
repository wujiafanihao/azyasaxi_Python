### 9.创建一个长度为10的随机数组并将最大值替换为-100

```python
import numpy as np

np.set_printoptions(suppress=True)
arr = np.random.randint(0,100,size=10)
print(arr)

arr_max = arr.max()
# 条件判断
cond = arr == arr_max
cond
arr[cond] = -100
arr
```

### 10.如何根据第三列大小顺序来对一个5*5矩阵排序? (考察（argsort()方法）)

```python
import numpy as np

np.random.seed(10)
arr = np.random.randint(0,10,(5,5))
print('orginal : \n',arr)

# 查看第三列数据,并获取索引排序
print('\n the third col database is : ',arr[:,2])
# argsort()方法返回的是数组值从小到大的索引值

print('\n the third col database sorted : ',np.argsort(arr[:,2])) # 返回的是索引值

# 根据第三列索引排序,进行重新索引
print('\nbase on the third col, Then reindex is : \n',arr[np.argsort(arr[:,2])])
```
