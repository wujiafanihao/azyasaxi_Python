### 9.创建一个长度为10的随机数组并将最大值替换为-100

```python
arr = np.random.randint(0,100,size=10,dtype=np.int16)
arr
```

```python
arr_max = arr.max()
arr_max
```

```python
cond = arr == arr_max
arr[cond] = -100
arr
```

### 10.如何根据第三列大小顺序来对一个5*5矩阵排序? (考察（argsort()方法）)

```python
arr = np.random.randint(1,10,(5,5))
arr
```

```python
print('the third col is : ', arr[:,2])
```

```python
sort = arr[:,2].argsort()
print('the third col is : ',sort )
```

```python
arr[sort]
```
