### 10.如何根据第三列大小顺序来对一个5*5矩阵排序? (考察（argsort()方法）)

```python
import numpy as np

np.random.seed(10)
arr = np.random.randint(0,10,(5,5))
print('orginal : \n',arr)

# 查看第三列数据,并获取索引排序
print('\n the third col database is : ',arr[:,2])
# argsort()方法返回的是数组值从小到大的索引值

print('\n the third col database sorted : ',np.argsort(arr[:,2])) # 返回的是索引值

# 根据第三列索引排序,进行重新索引
print('\nbase on the third col, Then reindex is : \n',arr[np.argsort(arr[:,2])])
```

### 计算每一行的softmax得分(exp表示自然底数e的幂运算)

```python
np.set_printoptions(suppress=True)
np.exp(2) # e的二次方
```

```python
def softmax(x):
    exp = np.exp(x)
    # print(exp)
    # print(exp.sum(axis = 1))
    # 每一行求和，并且进行形状改变（变成二维，可进行广播）
    # axis = 1 表示计算每一行的和
    result = (exp / exp.sum(axis = 1).reshape(-1,1)).round(1)
    return result
```
