```python
import random

li = []
for i in range(10000000):
    li.append(random.random())
```

```python
import time

t1 = time.time()
ret = sum(li)
t2 = time.time()
print(f"耗时间{t2-t1}")
print(ret)
```

```python
import numpy as np
n = np.array(li)

t1 = time.time()
ret = np.sum(n)
t2 = time.time()
print(f"耗时间{t2-t1}")
print(ret)
```

```python
#结论numpy数组处理数据别python列表快
```
