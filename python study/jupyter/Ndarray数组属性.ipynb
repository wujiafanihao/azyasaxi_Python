```python
import numpy as np

arr = np.arange(24)
print(arr)
print(arr.ndim)

#每一个元素有三行四列,有两个元素
arr2 = arr.reshape(2,3,4)
print(arr2)
```

```python
arr = np.array([[1,2,3],[4,5,6]])
print(arr)

print(arr.shape)
```

```python
arr = np.array([[1,2,3],[4,5,6]])
arr.shape = (3,2)
print(arr)
print(arr.shape)
```

```python
arr = np.array([[1,2,3],[4,5,6]])
print(arr)
print(arr.size)
print(arr.dtype)
```

```python
arr1 = np.array([1,2,3,4,5],dtype='i1')
print(arr1)
print(arr1.itemsize)

arr2 = np.array([1,2,4,5,6],dtype='f')
print(arr2)
print(arr2.itemsize) #字节
```

```python
arr = np.array([1,2,3,4,5])
print(arr.flags)
```
