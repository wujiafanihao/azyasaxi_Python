### **字典一些进阶用法**

```python
#1. 字典推导 (Dictionary Comprehension)
# 创建一个平方字典
squares = {x: x**2 for x in range(10)}
print('字典推导：',squares)

#2. 字典合并 (Dictionary Merging)
dict1 = {'a': 1, 'b': 2}
dict2 = {'b': 3, 'c': 4}
combined = dict1 | dict2
print('字典合并：',combined)

#3. 条件过滤 (Filtering with Conditions)
# 过滤出值为偶数的键值对
evens = {k: v for k, v in squares.items() if v % 2 == 0}
print('条件过滤：',evens)

#4. 深拷贝 (Deep Copy)
import copy
original = {'a': 1, 'b': {'c': 2}}
deep_copied = copy.deepcopy(original)
deep_copied['b']['c'] = 99
print("起初:", original)
print("深拷贝:", deep_copied)

#5. 获取所有键、值或键值对 (Keys, Values, Items)
data = {'a': 1, 'b': 2, 'c': 3}
print('所有键：',data.keys())
print('值：',data.values())
print('键值对：',data.items())

#6. 字典元素计数 (Count Occurrences)
from collections import Counter
data = ['apple', 'banana', 'apple', 'cherry', 'banana', 'cherry', 'cherry']
counter = Counter(data)
print('字典元素计数：',counter)

#7. 通过键查找值 (Lookup by Key)
data = {'a': 1, 'b': 2}
print('通过键查找值：',data.get('a'))  # 输出: 1
print('通过键查找值：',data.get('c', 0))  # 输出: 0（'c' 不存在，返回默认值 0）

#8. 遍历字典 (Iterate through Dictionary)
data = {'a': 1, 'b': 2, 'c': 3}
for key, value in data.items():
    print('遍历字典：',key, value)

#9. 默认字典 (Default Dictionary)
from collections import defaultdict
dd = defaultdict(int)
dd['a'] += 1
print(dd)  # 输出: defaultdict(<class 'int'>, {'a': 1})

#10. 字典排序 (Sorting Dictionary)
data = {'a': 3, 'b': 1, 'c': 2}
# 按键排序
sorted_by_key = dict(sorted(data.items()))
print(sorted_by_key)
# 按值排序
items = data.items()
print(items)
print(list(items))  # 输出: [('a', 3), ('b', 1), ('c', 2)]
sorted_items = sorted(items, key=lambda item: item[1])
print(sorted_items)  # 输出: [('b', 1), ('c', 2), ('a', 3)]
sorted_by_value = dict(sorted_items)
print(sorted_by_value)  # 输出: {'b': 1, 'c': 2, 'a': 3}
```

### **列表一些进阶的用法**

```python
#1. 列表推导 (List Comprehension)，列表推导是一种简洁的创建列表的方式。
# 创建一个平方列表
squares = [x**2 for x in range(10)]
print('列表推导：',squares)

#2. 列表合并 (List Concatenation)
arr1 = [1, 2, 3]
arr2 = [4, 5, 6]
combined = arr1 + arr2
print('列表合并：',combined)

#3. 条件过滤 (Filtering with Conditions)
# 过滤出偶数
evens = [x for x in range(10) if x % 2 == 0]
print('条件过滤：',evens)

#4. 深拷贝 (Deep Copy)
import copy
original = [1, 2, [3, 4]]
deep_copied = copy.deepcopy(original)
deep_copied[2][0] = 99 #索引为2的列表里面的第一个元素
print("起初:", original)
print("深拷贝:", deep_copied)

#5. 统计信息 (Statistics)
numbers = [1, 2, 3, 4, 5]
# 最大值
print('最大值：',max(numbers))
# 最小值
print('最小值：',min(numbers))
# 总和
print('总和：',sum(numbers))

#6. 使用 count 方法可以统计列表中某个元素出现的次数。
chars = ['a', 'b', 'a', 'c', 'a', 'b']
print('使用 count 方法可以统计列表中某个元素出现的次数：',chars.count('a'))  # 输出 3

#7. 查找元素索引 (Find Index)
fruits = ['apple', 'banana', 'cherry', 'apple']
print('查找元素索引：',fruits.index('apple'))  # 输出 0

#8. 通过条件生成列表 (Generate List by Condition)
# 将字符串转换为其ASCII值的列表
ascii_values = [ord(c) for c in "Hello"]
print("通过条件生成列表：",ascii_values)

#9. 展开嵌套列表 (Flatten Nested Lists)
nested_list = [[1, 2, 3], [4, 5], [6, 7, 8]]
flattened = [item for sublist in nested_list for item in sublist]
print('展开嵌套列表：',flattened)

#10. 使用 enumerate 遍历列表
#enumerate 函数可以同时获取元素的索引和值。
animals = ['cat', 'dog', 'bird']
for index, animal in enumerate(animals):
    print('索引：',index, '值：',animal)
```

```python
# 获取数据
data_array = data['data']
feature_names = data['feature_names']

# 打印原始数据
print("原始数据:")
for row in data_array:
    print(row)

# 过滤掉包含nan的行
# 我们使用列表推导式来创建一个新的列表 'filtered_data'，
# 其中包含原始数据数组 'data_array' 中不包含 nan 的行。
filtered_data = [
    row  # 对于原始数据数组中的每一行
    for row in data_array
    # 检查这一行是否包含任何 nan 值
    if not np.any([
        # 对于这一行中的每个元素
        element == 'nan' or element != element
        # 检查该元素是否等于 'nan'（字符串形式）或该元素不等于自身（这是 nan 的一个特性）
        for element in row #用于迭代 row 中的每一个元素
    ])
    # 如果该行不包含任何 nan 值，则将其包含在 'filtered_data' 中
]

# 打印过滤后的数据
print("\n过滤后的数据:")
for row in filtered_data:
    print(row)
```

```python
#使用属性

class Dog:
  def __init__(self,name,age,sex='磁性'):
    self.name = name
    self.__age = age
    self.sex = sex

  def run(self):
    print("{0}在跑...".format(self.name))

  @property #代替get_age,使用property装饰器进行修饰，方法名就是属性名就是age
  def age(self):
    return self.__age

  #访问私有变量先写一个公有变量方法
  def get_age(self):
    return self.__age
  #更新私有变量age参数
  def set_age(self,age):
    self.__age = age
    return self.__age

  @age.setter #代替set_age，setter装饰器=修改
  def age(self,age):
    self.__age = age

dog = Dog('球球',2)
print("The dog age is {0}".format(dog.age))
dog.age = 3
print("The dog age is {0}".format(dog.age))

print("The dog age is {0}".format(dog.get_age()))
print("The dog age is {0}".format(dog.set_age(4)))
```

### **继承性**

```python
#继承
class Animal:
  def __init__(self,name):
    self.name = name
  def show_info(self):
    return "动物的名字：{0}".format(self.name)
  def move(self):
    print("动一动")

class Cat(Animal):

  def __init__(self,name,age):
    super().__init__(name)
    self.age = age

cat = Cat('小猫',2)
cat.move()
print(cat.show_info())
```

```python
#多继承，同一种方法取到不同的值
class Horse:
  def __init__(self, name):
    self.name = name
  def show_info(self):
    return "the horse name is {0}".format(self.name)
  def run(self):
    print('the horse is running....')

class Donkey:
  def __init__(self, name):
    self.name = name
  def show_info(self):
    return "the donkey name is {0}".format(self.name)
  def run(self):
    print('the donkey is running....')
  def roll(self):
    print('the donkey is rolling....')

class Mule(Horse, Donkey):
  def __init__(self, name, age, is_horse=True):
    super().__init__(name)
    self.age = age
    self.is_horse = is_horse

  def show_info(self):
    if self.is_horse:
      return Horse.show_info(self)
    else:
      return Donkey.show_info(self)

mute1 = Mule('小马', 2, is_horse=True)
print("{0}, it age is {1}".format(mute1.show_info(), mute1.age))
mute1.run()

mute2 = Mule('小驴', 3, is_horse=False)
print("{0}, it age is {1}".format(mute2.show_info(), mute2.age))
mute2.roll()
```

### **继承与多态**

```python
class Animal:
  def speak(self):
    print("不知道是哪种动物再叫")

class Dog(Animal):
  def speak(self):
    print("狗再叫：汪汪汪")

class Cat(Animal):
  def speak(self):
    print("猫再叫：喵喵喵")

class Car(Animal):
  def speak(self):
    print("车再叫：嗡嗡嗡")

dog = Dog()
cat = Cat()
Animal.speak("鸡")
dog.speak()
cat.speak()

def start(obj): #接受obj对象，具有speak的方法
  obj.speak()

start(Car())
```

```python
#多个异常
i = input("input number :")
n = 8888
try:
  result = n / int(i)
  print(result)
  print("{0} / {1} = {2}".format(n,i,result))
except ZeroDivisionError as e:
  print("enable not to / zero, error：{0}".format(e))
except ValueError as e:
  print("input error, error：{0}".format(e))
except Exception as e: #总的
  print("error：{0}".format(e))
except:
  print("error")
```

### **datetime库**

```python
#datetime
import datetime

# 获取当前日期和时间
now = datetime.datetime.now()
local_time = datetime.datetime.today()
print(local_time)  # 输出本地时间
print(now)  # 输出当前日期和时间
#美国时间
print(datetime.datetime.utcnow())  # 输出UTC时间
#日本时间
print(datetime.datetime.now(datetime.timezone(datetime.timedelta(hours=9))))  # 输出日本时间

# 创建一个指定日期和时间的datetime对象
dt = datetime.datetime(2023, 6, 17, 7, 50, 48)
print(dt)  # 输出指定的日期和时间

# 将字符串转换为datetime对象
dt_str = "2023-06-17 07:50:48"
dt = datetime.datetime.strptime(dt_str, "%Y-%m-%d %H:%M:%S")
print(dt)  # 输出转换后的datetime对象

# 将datetime对象转换为字符串
dt_str = dt.strftime("%Y-%m-%d %H:%M:%S")
print(dt_str)  # 输出转换后的字符串

# 获取日期和时间的各个部分
year = dt.year
month = dt.month
day = dt.day
hour = dt.hour
minute = dt.minute
second = dt.second
print(year, month, day, hour, minute, second)  # 输出各个部分

# 日期和时间的加减
new_dt = dt + datetime.timedelta(days=1, hours=2)
print(new_dt)  # 输出加减后的日期和时间
```

```python
#datetime.date 类的方法示例
from datetime import date

# 获取今天的日期
today = date.today()
print(today)  # 输出：2024-06-17

# 创建一个指定日期
specific_date = date(2024, 6, 17)
print(specific_date)  # 输出：2024-06-17

# 获取日期的年份、月份和日期
year = specific_date.year
month = specific_date.month
day = specific_date.day
print(year, month, day)  # 输出：2024 6 17

# 获取日期的星期几（0是星期一，6是星期日）
weekday = specific_date.weekday()
print(weekday)  # 输出：0（表示星期一）

# 获取日期的ISO格式的星期几（1是星期一，7是星期日）
isoweekday = specific_date.isoweekday()
print(isoweekday)  # 输出：1（表示星期一）

# 获取日期的ISO日历（年份、星期数、星期几）
iso_calendar = specific_date.isocalendar()
print(iso_calendar)  # 输出：(2024, 25, 1)

# 将日期转换为ISO格式的字符串
iso_format = specific_date.isoformat()
print(iso_format)  # 输出：2024-06-17

# 将日期转换为字符串
date_str = specific_date.strftime("%Y-%m-%d")
print(date_str)  # 输出：2024-06-17
```

```python
#datetime.time 类的方法示例
from datetime import time

# 创建一个指定时间
specific_time = time(14, 30, 45)
print(specific_time)  # 输出：14:30:45

# 获取时间的小时、分钟、秒和微秒
hour = specific_time.hour
minute = specific_time.minute
second = specific_time.second
microsecond = specific_time.microsecond
print(hour, minute, second, microsecond)  # 输出：14 30 45 0

# 将时间转换为字符串
time_str = specific_time.strftime("%H:%M:%S")
print(time_str)  # 输出：14:30:45
```

```python
#datetime.timedelta 类的方法示例
from datetime import timedelta

# 创建一个时间间隔
delta = timedelta(days=5, hours=3, minutes=30)
print(delta)  # 输出：5 days, 3:30:00

# 获取时间间隔的天数、秒数和微秒数
days = delta.days
seconds = delta.seconds
microseconds = delta.microseconds
print(days, seconds, microseconds)  # 输出：5 12600 0

# 两个日期相减得到时间间隔
date1 = date(2024, 6, 17)
date2 = date(2024, 6, 10)
delta = date1 - date2
print(delta)  # 输出：7 days, 0:00:00

# 时间间隔的加减运算
new_date = date1 + delta
print(new_date)  # 输出：2024-06-24
```

```python
#datetime.tzinfo 类的方法示例
from datetime import tzinfo, timedelta, datetime

class SimpleTZ(tzinfo):
    def __init__(self, offset_hours, name):
        self.offset = timedelta(hours=offset_hours)
        self.name = name

    def utcoffset(self, dt):
        return self.offset

    def tzname(self, dt):
        return self.name

    def dst(self, dt):
        return timedelta(0)

# 创建一个时区对象
simple_tz = SimpleTZ(8, "CST")

# 获取当前时间并设置时区
now = datetime.now(simple_tz)
print(now)  # 输出：2024-06-17 07:57:11+08:00

# 获取时区的UTC偏移量
offset = simple_tz.utcoffset(now)
print(offset)  # 输出：8:00:00

# 获取时区的名称
tz_name = simple_tz.tzname(now)
print(tz_name)  # 输出：CST
```

## **自动化搜索文件**

```python
import os #系统库os
import shutil #实现移动文件的功能

#将path目录下的文件保存到result
def search_dir(path,result): #path代表待搜索的目录路径，result等于搜索到的文件路径
  child_files = os.listdir(path)
  for child in child_files:
    child = os.path.join(path,child)

    result.append(child)
    if os.path.isdir(child): #如果child是一个子目录
      search_dir(child,result)

if __name__ == "__main__":
  path = input("请输入待搜索的目录路径：")
  out_path = input("请输入输出路径：")
  result = []
  search_dir(path,result)
  print("搜索到的文件为：")
  for item in result:
    print(item)
    if os.path.isfile(item) and item.endswith('.xlsx'):
      shutil.copy(item,out_path)
      print(f"{item}复制成功")
```

```python
import urllib.request
from bs4 import BeautifulSoup

# Function to download HTML content from a URL
def download_html(url):
    header = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/125.0.0.0 Safari/537.36'
    }
    req = urllib.request.Request(url=url, headers=header)
    response = urllib.request.urlopen(req)
    html = response.read().decode("utf-8")
    return html

# Example URL
url = "https://movie.douban.com/top250"

# Download HTML content for the example URL
html = download_html(url)
print(html)
```
