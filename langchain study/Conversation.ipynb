```python
def track_tokens_usage(chain,query):
    with get_openai_callback() as cb:
        result = chain.invoke(query)
        print(f"token usage：{cb.total_tokens}")
        print(f"token cost：${cb.total_cost:.12f}")

    return result
```

```python
llm = ChatOpenAI(temperature=0,base_url=BASE_URL,api_key=API_KEY,model="gpt-3.5-turbo")

track_tokens_usage(llm,"What is langchain?").content
```

```python
conversation = ConversationChain(llm=llm,memory=ConversationBufferMemory())
print(conversation.prompt.template)
```

```python
track_tokens_usage(conversation,"What is langchain?")['response']
```

```python
track_tokens_usage(conversation,"What did I just say?")['response']
```
