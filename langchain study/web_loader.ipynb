```python
from langchain_community.document_loaders.web_base import WebBaseLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain.prompts import ChatPromptTemplate,SystemMessagePromptTemplate,HumanMessagePromptTemplate
from langchain_openai import ChatOpenAI
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
base_url = os.getenv("OPENAI_BASE_URL")
openai_api_key = os.getenv("OPENAI_API_KEY")

print(base_url)
print(openai_api_key)
```

###网页总结

```python
loader = WebBaseLoader('https://mp.weixin.qq.com/s/Stq_XIEM2Y0dM87vbjcjSQ')
docs = loader.load()
# print(docs)
```

```python
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000,chunk_overlap=200)
splitter = text_splitter.split_documents(docs)
print(len(splitter))
# print(splitter)
```

```python
SAM_prompt = ChatPromptTemplate.from_messages([
    SystemMessagePromptTemplate.from_template("According to user's requirements To summarize this context :{context} and then returns it to the user"),
    HumanMessagePromptTemplate.from_template("{requirements}")
])
llm = ChatOpenAI(temperature=0,verbose=True)
SAM_chain = SAM_prompt | llm
```

```python
response = SAM_chain.invoke(input={
    "context": docs,
    "requirements": "Humanity,authenticity,Answer in Chinese and 800 word summary",
})
print(f"SUMMARIZE ：{response.content}")
```

```python
response = SAM_chain.invoke(input={
    "context": docs,
    "requirements": "Answer in Chinese and 400 word summary",
})
print(f"SUMMARIZE ：{response.content}")
```
