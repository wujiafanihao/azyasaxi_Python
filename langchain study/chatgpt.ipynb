```python
from langchain_core.prompts import ChatPromptTemplate,SystemMessagePromptTemplate,HumanMessagePromptTemplate,MessagesPlaceholder

prompt = ChatPromptTemplate.from_messages([
    SystemMessagePromptTemplate.from_template("answer the user's question"),
    MessagesPlaceholder(variable_name="chat_history"),
    HumanMessagePromptTemplate.from_template("{question}"),
])
```

```python
from langchain.memory import ConversationBufferWindowMemory
from langchain.chains.conversation.base import LLMChain

memory = ConversationBufferWindowMemory(return_messages=True,memory_key="chat_history",k=3)
llm_chain = LLMChain(
    llm=llm,
    memory=memory,
    prompt=prompt,
    # verbose=True
)
```

```python
query = "我问你假如，有一个网站链接结构是https://movie.douban.com/subject/34937650/celebrities，他那个34937650是每个影片的id，他没有规律是随机的，那我该怎样批量获取这个页面的内容呢？"
llm_chain.invoke(input={"question":query})
```
