# **任务 1：数据读取与处理**

## **任务 1.1 数据读取**

```python
import pandas as pd

user = pd.read_csv('data/user.csv')
print('user：',user)

author = pd.read_csv('data/author.csv')
print('author：',author)

work_info = pd.read_csv('data/item.txt')
print('work_info：',work_info)
```

```python
#保存为变量
user.to_csv('data/df_user.csv',index = True)
author.to_csv('data/df_author.csv', index = True)
work_info.to_csv('data/df_item.csv', index = True)
print('完成')
```

## **任务 1.2 user 表处理**

### **任务 1.2.1 无效列处**

```python
user_data = pd.read_csv('data/df_user.csv')

#删除列字段`Unnamed: 0`
if 'Unnamed: 0'in user_data.columns:
    user_data.drop(columns = ['Unnamed: 0'], inplace = True)
else:
    print('没找到 Unnamed: 0')

user_data.to_csv('data/df_user.csv')
print('已删除')
```

```python
print(user_data)
```

### **任务 1.2.2 字段类型修改**

```python
#查看原始类型
print(user_data.dtypes)
```

```python
#修改列字段`uid`和`user_city`的数据类型为`object`
user_data['uid'] = user_data['uid'].astype('object')
user_data['user_city'] = user_data['user_city'].astype('object')

user_data.to_csv('data/df_user.csv')
print('修改为object成功')
```

```python
#查看是否修改成功
print(user_data.dtypes)
```

## **任务 1.3 author 表处**

### **任务 1.3.1 重复值处理**

```python
author_data = pd.read_csv('data/df_author.csv')

# 找到重复的行
duplicates = author_data[author_data.duplicated()]

# 循环打印删除的元素
for index, row in duplicates.iterrows():
    print(f"删除的元素: {row.to_dict()}")

# 删除多余行重复值
author_data.drop_duplicates(inplace=True)

author_data.to_csv('data/df_author.csv')
print('已删除完成')
```

```python
print(author_data)
```

### **任务 1.3.2 重命名列名**

```python
#将字段`id`重命名为`author_id`
author_data.rename(columns={'id': 'author_id'}, inplace=True)
print('修改完成')
```

```python
print(author_data)
```

## **任务 1.4 item 表处**

###  **任务 1.4.1 空值处理**

```python
work_info_data = pd.read_csv('data/df_item.csv')

# 删除存在空值的行数据
work_info_data.dropna(inplace=True)

work_info_data.to_csv('data/df_item.csv')
print('已完成')
```

```python
print(work_info_data)
```

### **任务 1.4.2 重置索引**

```python
# 重置索引，并删除之前的索引
work_info_data.reset_index(drop=True, inplace=True)

work_info_data.to_csv('data/df_item.csv')
print('已删除')
```

```python
print(work_info_data)
```

### **任务 1.4.3 时间字段处理**

```python
#提取 `real_time` 字段的日期信息，保存到新增列 `date`
work_info_data['date'] = pd.to_datetime(work_info_data['real_time']).dt.date

# 提取 `real_time` 字段的小时段信息，保存到新增列 `hour`
work_info_data['hour'] = pd.to_datetime(work_info_data['real_time']).dt.hour

work_info_data.to_csv('data/df_item.csv')
print('已完成')
```

```python
print(work_info_data)
```

## **任务 1.5 数据合并**

```python
# 合并 user_data 和 author_data
df_user_author = pd.merge(user_data, author_data, on='item_id', how='inner')

# 合并 df_user_author 和 work_info_data
df = pd.merge(df_user_author, work_info_data, on='item_id', how='inner')

df.to_csv('data/df.csv', index=False, sep=', ')

print('完成')
```
