```python
from langchain_openai import AzureChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain.chains.llm import LLMChain
from dotenv import load_dotenv
import pandas as pd
import os
import openpyxl

load_dotenv()

llm = AzureChatOpenAI(
    temperature=0,
    azure_deployment='tcl-gpt4o1'
)

template = """
You are the TV test corpus writing assistant, 
and you will be writing based on the movie and TV material I gave you: {context}, 
written in groups of three to four sentences.

here is an example of the formatting:
<pre>
播放琅琊榜
讲的是什么
梅长苏是谁饰演的

播放轩辕剑
胡歌在里面饰演的是什么
这部剧讲的是什么

请播放热辣滚烫
导演是谁
主演是谁
他的代表作有什么

播放飞驰人生
范丞丞在里面饰演什么
他的代表作
2024的

请播放第二十条
导演是谁
他的还有什么代表作吗
我想看易烊千玺出演的

播放抓娃娃
这部电影讲的是什么
沈腾在里面饰演什么
快进10分钟

我想看2024九龙城寨之围城
古天乐在里面饰演什么
陈洛军在里面是谁饰演
他的代表作

2024年胡歌演的最新电影
和高圆圆演的哪一部
这部电影讲的是什么
导演是谁
</pre>

Just generate one group at a time.
don't use markdown code.
Don't generate example of the formatting for me.
All you need to do is generate the corpus of the movie or television I provided you with, as well as his background.
"""

def init_model(context):
    prompt = PromptTemplate(
        template=template,
        input_variables=['context']
    )
    llm_chain = LLMChain(
        llm=llm,
        prompt=prompt,
    )
    res = llm_chain.invoke({
        'context': context,
    })
    return res

def generate_q(file_path):
    df = pd.read_excel(file_path)
    a_column = df['A'].tolist()
    return a_column

def append_to_excel(new_data, output_path):
    if os.path.exists(output_path):
        # Load existing workbook and get the active sheet
        book = openpyxl.load_workbook(output_path)
        sheet = book.active
        
        # Find the last row and start appending data
        last_row = sheet.max_row
        for row in new_data:
            sheet.cell(row=last_row + 1, column=1, value=row)
            last_row += 1
        book.save(output_path)
    else:
        # Create new workbook and add data
        book = openpyxl.Workbook()
        sheet = book.active
        for row in new_data:
            sheet.append([row])
        book.save(output_path)

def main(file_path, output_path):
    q_generate = generate_q(file_path)

    for q in q_generate:
        res = init_model(q)
        print(res['text'])
        # Prepare the result with each sentence as a separate row
        result = res['text'].split('\n')
        result.append("")  # Add a blank line to separate groups
        
        # Append result to the Excel file
        append_to_excel(result, output_path)

main(file_path='merged_douban_actor.xlsx', output_path='output_results.xlsx')
```
