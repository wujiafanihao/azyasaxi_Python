```python
def random_sample_with_delimiter(df, delimiter_column, num_samples):
    # 找到NaN值的索引
    nan_indices = df[df[delimiter_column].isna()].index.tolist()

    # 根据这些索引将数据框分成多个组，包括分界线行
    groups = []
    start_idx = 0
    for idx in nan_indices:
        groups.append(df.iloc[start_idx:idx+1])  # 包括分界线行
        start_idx = idx + 1
    groups.append(df.iloc[start_idx:])  # 添加最后一个组

    # 移除空组
    groups = [group for group in groups if not group.empty]

    # 随机抽取指定数量的组
    sampled_groups = random.sample(groups, min(num_samples, len(groups)))

    # 将抽取的组合并成一个新的数据框
    sampled_df = pd.concat(sampled_groups, ignore_index=True)

    return sampled_df
```

```python
mul_conversation_1__df = random_sample_with_delimiter(df_4, '语料', 40)
mul_conversation_2__df = random_sample_with_delimiter(df_5, '语料', 40)
mul_conversation_3__df = random_sample_with_delimiter(df_6, '语料', 40)

print(mul_conversation_1__df)
```

```python
# 随机抽取数据
def random_sample_within_range(group, min_samples, max_samples):
    num_samples = random.randint(min_samples, max_samples)
    return group.sample(n=min(num_samples, len(group)), random_state=1)
```

```python
# 按分类标签分组，并在每个组内随机抽取数据
min_samples = 10
max_samples = 30 
Intent_comprehension_df = df_1.groupby('分类').apply(lambda x: random_sample_within_range(x, min_samples, max_samples)).reset_index(drop=True)
Intent_Bootstrapping_df = df_2.groupby('分类').apply(lambda x: random_sample_within_range(x, min_samples, max_samples)).reset_index(drop=True)
Entity_extraction_df = df_3.groupby('分类').apply(lambda x: random_sample_within_range(x, min_samples, max_samples)).reset_index(drop=True)
security_df = df_7.groupby('分类').apply(lambda x: random_sample_within_range(x, min_samples, max_samples)).reset_index(drop=True)
Semantic_rejection_df = df_8.groupby('分类').apply(lambda x: random_sample_within_range(x, min_samples, max_samples)).reset_index(drop=True)
Content_Search_df = df_9.groupby('分类').apply(lambda x: random_sample_within_range(x, min_samples, max_samples)).reset_index(drop=True)
Complex_tasks_df = df_10.groupby('分类').apply(lambda x: random_sample_within_range(x, min_samples, max_samples)).reset_index(drop=True)
natural_chat_df = df_11.groupby('分类').apply(lambda x: random_sample_within_range(x, min_samples, max_samples)).reset_index(drop=True)
Diversity_of_discourse_df = df_12.groupby('分类').apply(lambda x: random_sample_within_range(x, min_samples, max_samples)).reset_index(drop=True)
```
